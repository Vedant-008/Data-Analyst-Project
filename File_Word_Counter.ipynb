```
from collections import Counter
import re
import sys

def analyze_file(file_path):
  try:
    with open(file_path, 'r', encoding='utf-8') as file:
      text = file.read()

    # Calculate basic metrics
    lines = text.splitlines()
    line_count = len(lines)
    char_count_with_spaces = len(text)
    char_count_no_spaces = len(text.replace(' ', '').replace('\n', ''))

    # Extract words (ignoring punctuation and case)
    words = re.findall(r'\b\w+\b', text.lower())
    word_count = len(words)

    # Get top 5 most common words
    most_common = Counter(words).most_common(5)

    # Display Results
    print('=' * 40)
    print(f'FILE ANALYSIS REPORT: {file_path}')
    print('=' * 40)
    print(f'Total Words:                  {word_count}')
    print(f'Total Lines:                  {line_count}')
    print(f'Characters (with spaces):     {char_count_with_spaces}')
    print(f'Characters (without spaces):  {char_count_no_spaces}')
    print('-' * 40)
    print('Top 5 Most Frequent Words:')
    for word, count in most_common:
      print(f"  - '{word}': {count} times")
    print('=' * 40)

  except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
  except UnicodeDecodeError:
    print(
        f"Error: Could not decode '{file_path}'. Please ensure it is a valid"
        ' UTF-8 text file.'
    )
  except Exception as e:
    print(f'An unexpected error occurred: {e}')

# --- Create a sample file and analyze it without interactive input ---
if __name__ == '__main__':
  sample_path = 'sample.txt'

  # Creating a dummy text file for demonstration
  with open(sample_path, 'w', encoding='utf-8') as f:
    f.write('Hello world!\nThis is a sample text file for testing the word counter.\nHello again!')

  # Analyze the sample file directly
  analyze_file(sample_path)
```
